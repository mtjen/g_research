# Data Clean: Survey

```python
survey_data = dataset_99672230_survey_df
survey_data.head()
```

```python
# change survey date
survey_data['date'] = pandas.to_datetime(survey_data['survey_datetime']).dt.date
survey_data = survey_data.drop('survey_datetime', axis = 1)
survey_data.head()
```

```python
# get questions for overall health
for question in survey_data['question'].unique():
    if question.startswith('Overall Health') == True:
        print(question)
```

```python
# get current status and overall variables
curr_data = survey_data[survey_data['question'].str.endswith('Currently')]
ovr_data = survey_data[survey_data['question'].isin(['Overall Health: General Quality',
                                                        'Overall Health: General Physical Health',
                                                        'Overall Health: General Mental Health',
                                                        'Overall Health: General Social',
                                                        'Overall Health: General Health'])]

curr_data = curr_data.reset_index().drop('index', axis = 1)
ovr_data = ovr_data.reset_index().drop('index', axis = 1)
```

## Data Clean: Survey: Current Status

```python
# fix skipped answers - current status variables
print(len(curr_data['answer'].unique()))
new_vals = []


for row_index in range(curr_data.shape[0]):
    ans_response = curr_data['answer'][row_index]
    # check to see if skipped
    if ans_response == 'PMI: Skip':
        question_response = curr_data['question'][row_index]
        split_vals = question_response.split(":")
        question = split_vals[1].strip()
        ans_response = question + ': No'
    new_vals.append(ans_response)
    
    
curr_data['new_vals'] = new_vals
print(len(curr_data['new_vals'].unique()))

curr_data.head()
```

```python
sorted(curr_data['new_vals'].unique())
```

```python
# fix two mismatched question-answer pairs

for row_index in range(curr_data.shape[0]):
    ans_response = curr_data['new_vals'][row_index]
    # check to see if one of the two bad coded questions
    if ans_response == 'Other Digestive Condition Currently: No':
        curr_data['new_vals'][row_index] = 'Other Digestive Conditions Currently: No'
    elif ans_response == 'Alcohol Disorder Currently: No':
        curr_data['new_vals'][row_index] = 'Alcohol Or Drug Currently: No'
        
print(len(curr_data['new_vals'].unique()))
curr_data.head()
```

```python
curr_data['answer'] = curr_data['new_vals']
curr_data = curr_data.drop('new_vals', axis = 1)
curr_data.head()
```

```python
# get question and answer values from 'answer'
question_array = []
response_array = []

for row_index in range(0, curr_data.shape[0]):
    ans_response = curr_data['answer'][row_index]
    
    # split values on colon, get question (and remove 'currently') and response values
    split_vals = ans_response.split(":")
    question = split_vals[0][:-10] 
    response = split_vals[1].strip()
    
    # append values
    question_array.append(question)
    response_array.append(response)
    
# add variables to dataframe
curr_data['question'] = question_array
curr_data['response'] = response_array
curr_data = curr_data.drop(['question_concept_id', 'answer_concept_id', 'answer'], axis = 1)
curr_data.head()
```

## Data Clean: Survey: Overall

```python
sorted(ovr_data['answer'].unique())
```

```python
# fix skipped answers - overall variables
print(len(ovr_data['answer'].unique()))
new_vals = []

for row_index in range(ovr_data.shape[0]):
    ans_response = ovr_data['answer'][row_index]
    # check to see if mistyped or skipped
    if ans_response == 'General Mental Health: Excllent':
        ans_response = 'General Mental Health: Excellent'
    elif ans_response == 'PMI: Skip':
        question_response = ovr_data['question'][row_index]
        split_vals = question_response.split(":")
        question = split_vals[1].strip()
        ans_response = question + ': Unknown'
    new_vals.append(ans_response)
    
    
ovr_data['new_vals'] = new_vals
print(len(ovr_data['new_vals'].unique()))

sorted(ovr_data['new_vals'].unique())
```

```python
# get question and answer values from 'answer'
question_array = []
response_array = []

for row_index in range(0, ovr_data.shape[0]):
    ans_response = ovr_data['new_vals'][row_index]
    
    # split values on colon, get question (and remove 'currently') and response values
    split_vals = ans_response.split(":")
    question = split_vals[0]
    response = split_vals[1].strip()
    
    # append values
    question_array.append(question)
    response_array.append(response)
    
# add variables to dataframe
ovr_data['question'] = question_array
ovr_data['response'] = response_array
ovr_data = ovr_data.drop(['question_concept_id', 'answer_concept_id', 'answer'], axis = 1)
ovr_data.head()
```

```python
merged_age = merged_data

age_years = []

for row_index in range(merged_data.shape[0]):
    age_days = merged_data['age_days'][row_index]
    num_years = math.floor(age_days / 365)
    age_years.append(num_years)
    
merged_age['age_years'] = age_years
merged_age.head()
```
